```python
import os
import sys
import glob
import boto3
```

```python
# Import the S3 helpers from s3-api/awsapi.py (this notebook sits at the repo root).

module_path = os.path.abspath('s3-api')
if module_path not in sys.path:
    sys.path.append(module_path)

from awsapi import upload_file, list_objects
```

```python
# Upload the author corpora in data/corpora/ to the S3 bucket named by $S3_BUCKET, under data/

bucket_name = os.environ.get('S3_BUCKET')
if not bucket_name:
    raise RuntimeError('Set S3_BUCKET to the target bucket name before running this notebook')
remote_dir  = 'data/'
in_path = os.path.join('data', 'corpora') + '/'

for doc in glob.glob(in_path+'*.txt'):
    
    split_doc = doc.split('/')
    local_dir = '/'.join(split_doc[:-1]) + '/'
    fn = split_doc[-1]
    
    upload_file(bucket_name, local_dir, fn, remote_dir)
```

```python
files= list_objects(bucket_name, 'data')
print(files)
```
